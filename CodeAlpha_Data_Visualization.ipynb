```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Product": ["Mobile", "Laptop", "Tablet", "Headphones", "Watch"],
    "Sales": [50000, 80000, 30000, 20000, 40000]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
plt.bar(df["Product"], df["Sales"])
plt.title("Product Sales Analysis")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
plt.pie(
    df["Sales"],
    labels=df["Product"],
    autopct="%1.1f%%"
)
plt.title("Sales Distribution")
plt.show()
```

Key Findings
1. Highest Sales: The laptop recorded the highest sales of ₹80,000, making it the best-performing product in the dataset.
2. Lowest Sales: Headphones recorded the lowest sales of ₹20,000, indicating comparatively lower sales than the other products.
3. Total Sales: The total sales of all five products amounted to ₹2,20,000.
4. Product Comparison: The bar chart clearly compares the sales of Mobile, Laptop, Tablet, Headphones, and Watch, making it easy to identify the products with the highest and lowest sales.
5. Sales Distribution: The pie chart illustrates the percentage contribution of each product to total sales. Laptops contributed approximately 36.36%, while headphones contributed approximately 9.09%.
6. Business Insights: The results help identify popular products and products with lower sales. This information can help businesses plan inventory, improve marketing strategies, and make better decisions.

##Conclusion
The Data Visualization task successfully represents product sales using bar and pie charts. The bar chart helps compare sales figures across different products, while the pie chart shows each product's contribution to total sales.
From the analysis, laptops generated the highest sales, whereas headphones generated the lowest sales. These visual representations make complex data easier to understand and help identify important patterns and differences.
